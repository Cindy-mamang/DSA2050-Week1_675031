```python
import pandas as pd
# Load the retail dataset

df = pd.read_csv("retail_csv.csv")
print("Dataset loaded successfully!")
```

```python
# Display the first 5 rows of the dataset
df.head()

#shape of the dataset
df.shape

#dattatypeS
df.info()
    
```

```python
#Total revenue KPI 1
total_revenue = df["Revenue_KES"].sum()

print("Total Revenue: KES", total_revenue)
```

```python
#Number of transactions KPI 2
number_of_transactions = df["OrderID"].nunique()

print("Number of Transactions:", number_of_transactions)
```

```python
#Average transaction value KPI 3
average_transaction_value = total_revenue / number_of_transactions

print("Average Transaction Value: KES", round(average_transaction_value, 2))
```

```python
# Display the three KPIs together
print("RETAIL BUSINESS KPIs")
print("--------------------")
print("Total Revenue: KES", total_revenue)
print("Number of Transactions:", number_of_transactions)
print("Average Transaction Value: KES", round(average_transaction_value, 2))
```

```python
# Calculate total revenue for each product category
category_revenue = df.groupby("Category")["Revenue_KES"].sum()

category_revenue
```

```python
import matplotlib.pyplot as plt

# Create a bar chart showing revenue by category
category_revenue.plot(kind="bar")

plt.title("Total Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Revenue (KES)")
plt.xticks(rotation=0)
plt.show()
```

```python
### Interpretation

#Electronics generated the highest revenue at KES 934,500, followed closely by Personal Care at KES 931,800. Accessories generated the lowest revenue at KES 422,250
```

```python
# Calculate total revenue for each region
region_revenue = df.groupby("Region")["Revenue_KES"].sum()

region_revenue
```

```python
# Create a bar chart showing revenue by region
region_revenue.plot(kind="bar")

plt.title("Total Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Revenue (KES)")
plt.xticks(rotation=0)
plt.show()
```

```python
### Interpretation

#Kisumu generated the highest revenue at KES 878,050, while Nairobi generated the lowest revenue at KES 757,950 among the four regions.
```

```python
## Conclusion

#The analysis shows that the 300 retail transactions generated a total revenue of KES 3,206,450, with an average transaction value of KES 10,688.17. Electronics generated the highest revenue among the product categories, while Kisumu generated the highest revenue among the regions. The analysis provides a descriptive view of sales performance, but it cannot explain the reasons behind differences in sales or determine business profit because cost information is not included.
```
